```python
import numpy as np
from pymatching import Matching
```

```python
matching = Matching()
matching.add_edge(0, 1, weight=1)
matching.add_boundary_edge(0, fault_ids=0, weight=1)
matching.add_boundary_edge(1, weight=1)

matching.SO_calculator_setup()
matching.add_boundary_node_SO(np.uint(2))
matching.add_boundary_node_SO(np.uint(3))
matching.add_boundary_edge_SO(np.uint(0), np.uint(2))
matching.add_boundary_edge_SO(np.uint(1), np.uint(3))
matching.add_cycle_endpoints_pair_SO(np.uint(2), np.uint(3))

matching.draw()
```

```python
shots = np.array([[0, 0]])
matching.decode_batch_soft_output(shots)
```
